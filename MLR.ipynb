```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Fuel_Type'] = le.fit_transform(df['Fuel_Type'])
```

```python
x= df.drop('Price',axis =1)
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x, y,test_size = 0.2,random_state = 50)
```

```python

from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)

pred = lr.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test,pred)
```

```python

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
x_sta_train = sc.fit_transform(x_train)
x_sta_test = sc.transform(x_test)
lr = LinearRegression()
lr.fit(x_sta_train,y_train)

pred = lr.predict(x_sta_test)

from sklearn.metrics import r2_score
r2_score(y_test,pred)
```

```python
from sklearn.linear_model import Lasso
ls = Lasso()
ls.fit(x_train,y_train)
pred = ls.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test,pred)
```

```python
from sklearn.linear_model import Ridge
ls = Ridge(alpha=0.5,max_iter=2000)
ls.fit(x_train,y_train)
pred = ls.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test,pred)
```

Normalization and Standardization are feature scaling techniques. Normalization scales data between 0 and 1 using Min-Max scaling and is commonly used for distance-based algorithms like KNN and K-Means. Standardization transforms data to have a mean of 0 and a standard deviation of 1 using Z-score scaling, making it suitable for algorithms like Linear Regression, Logistic Regression, SVM, and PCA. Feature scaling improves convergence speed, prevents large-valued features from dominating, and often improves model accuracy.

Multicollinearity occurs when two or more independent variables are highly correlated, making coefficient estimates unstable and difficult to interpret. It can be detected using a correlation matrix or Variance Inflation Factor (VIF). Common techniques to address multicollinearity include removing highly correlated features, applying feature selection methods, using PCA to create uncorrelated components, or using regularization techniques such as Ridge Regression or Lasso Regression. Ridge reduces the impact of multicollinearity by shrinking coefficients, while Lasso can also eliminate less important features by setting their coefficients to zero.
