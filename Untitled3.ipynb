```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.boxplot()
```

```python
for i in df.select_dtypes(include=['int','float']).columns:
  q1 = df[i].quantile(0.25)
  q3 = df[i].quantile(0.75)
  iqr = q3 - q1
  lower = q1 - 1.5*iqr
  upper = q3 + 1.5*iqr

  df[i] = df[i].clip(lower,upper)

```

```python
x = df.drop(columns='Type')
y = df['Type']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=32)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
sd = StandardScaler()
x_train_s = sd.fit_transform(x_train)
x_test_s = sd.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
rc = RandomForestClassifier()
rc.fit(x_train_s,y_train)
```

```python
pred = rc.predict(x_test_s)
```

```python
from sklearn.metrics import accuracy_score,classification_report
print(accuracy_score(y_test,pred))
print(classification_report(y_test,pred))
```

```python
new_s = sd.transform(new)
```

```python
prediction = rc.predict(new_s)
print(prediction)
```

```python
from sklearn.ensemble import RandomForestClassifier

bagging = RandomForestClassifier(n_estimators=100,max_samples=0.5,
                                 max_features=0.3,max_depth=10)
bagging.fit(x_train_s,y_train)

pred = bagging.predict(x_test_s)

print(accuracy_score(y_test,pred))
```

```python
from sklearn.model_selection import GridSearchCV
grid = {
    'n_estimators': [50,100,200],
    'max_samples': [0.4,0.5,0.6,0.7],
    'max_features': [0.4,0.5,0.6,0.7]
}

random = RandomForestClassifier(max_depth=10,random_state=32)

grid_cv = GridSearchCV(estimator=random, param_grid=grid, cv=5,scoring='accuracy')

grid_cv.fit(x_train,y_train)

print(grid_cv.best_params_)
print(grid_cv.best_score_)
```
